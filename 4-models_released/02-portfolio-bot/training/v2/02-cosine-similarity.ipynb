```python
import numpy as np
```

```python
# Compute the cosine similarity between token 2 and token 5
# cos(θ) = (E[2] @ E[5]) / (||E[2]|| * ||E[5]||)
# If close to 1 → similar, close to 0 → independent
```

```python
E = np.random.randn(10, 4)
```

```python
magnitude_a = np.linalg.norm(E[2])
print(magnitude_a)
magnitude_b = np.linalg.norm(E[5])
print(magnitude_b)
```

```python
cos = (E[2] @ E[5]) / (magnitude_a * magnitude_b)
```

```python
print(cos)
```

```python
tokens = [3, 4, 6]
print(E[[tokens]])
```

```python
phrase_embeddings = E[tokens]
phrase_vector = phrase_embeddings.mean(axis=0)

print(phrase_embeddings)
print(phrase_vector)
print(phrase_vector.shape)
```

```python
result = []
query = np.array([0.5, -0.2, 0.1, 0.8])
for e in tokens:
    magnitude_a = np.linalg.norm(query)
    magnitude_b = np.linalg.norm(E[e])
    cos = (query @ E[e]) / (magnitude_a * magnitude_b)
    result.append(cos)
```

```python
cos_result = np.array(result)
print(result)
print(f"token most similar : {np.argmax(result)}")
```

```python
# transformer
norms = np.linalg.norm(E, axis=1)
dots = E @ query
similarity = dots / (norms * np.linalg.norm(query))

print(similarity)
print(f"Most similar token {np.argmax(similarity)}")
```
