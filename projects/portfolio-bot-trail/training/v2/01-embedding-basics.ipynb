```python
import numpy as np

E = np.random.randn(10, 4)
print(E)
```

```python
def embed(indices):
    return(E[indices])
```

```python
result = embed([2, 3, 3])
print(result)
print(result.mean(axis=0))
```

```python
one_hot = np.zeros(10)
one_hot[5] = 1
print(one_hot)
print(one_hot @ E)
print(E[5])
```
